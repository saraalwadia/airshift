### Missing Value

```python
missing = sample_df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

missing
```

```python
missing_percentage = (
    sample_df.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage[missing_percentage > 0]
```
