## 2. Cross-Station Profiling

## 3. Temporal Profiling

Since AirShift is based on hourly time-series data, we first verify the temporal consistency of all monitoring stations before handling missing values.

### 3.1 Timestamp Consistency

```python
temporal_summary = []

for file in files:
    df = pd.read_csv(file)

    df["datetime"] = pd.to_datetime(
        df[["year", "month", "day", "hour"]]
    )

    temporal_summary.append({
        "station": df["station"].iloc[0],
        "start": df["datetime"].min(),
        "end": df["datetime"].max(),
        "rows": len(df),
        "unique_timestamps": df["datetime"].nunique(),
        "duplicate_timestamps": df["datetime"].duplicated().sum(),
        "chronologically_sorted": df["datetime"].is_monotonic_increasing
    })

temporal_df = pd.DataFrame(temporal_summary)

temporal_df
```

### Temporal Profiling Findings

All 12 stations have consistent temporal coverage from **March 2013 to February 2017** with 35,064 unique hourly timestamps each. No duplicate timestamps were detected, and all datasets are chronologically ordered.

The next step is to check for **missing hourly timestamps (time gaps)**.


### 3.2 Time Gap Analysis

We check for missing hourly timestamps to ensure the time series is continuous across all monitoring stations.

```python
gap_summary = []

for file in files:
    df = pd.read_csv(file)

    df["datetime"] = pd.to_datetime(
        df[["year", "month", "day", "hour"]]
    )

    time_diff = df["datetime"].diff()

    gaps = (time_diff > pd.Timedelta(hours=1)).sum()

    gap_summary.append({
        "station": df["station"].iloc[0],
        "time_gaps": gaps,
        "largest_gap": time_diff.max()
    })

gap_df = pd.DataFrame(gap_summary)

gap_df
```
